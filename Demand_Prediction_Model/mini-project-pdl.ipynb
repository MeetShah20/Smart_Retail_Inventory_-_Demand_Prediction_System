```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Load the dataset
file_path = '/kaggle/input/super-store-sales-data-dataset/Super-Store-Sales-Data_.csv'
df = pd.read_csv(file_path, encoding='ISO-8859-1')

# 2. Enhanced Exploratory Data Analysis Functions

def get_descriptive_statistics(df, numerical_cols):
    """
    Generates descriptive statistics for numerical columns.

    Args:
        df (pd.DataFrame): The input DataFrame.
        numerical_cols (list): List of numerical columns.

    Returns:
        pd.DataFrame: Descriptive statistics.
    """
    return df[numerical_cols].describe()

def analyze_categorical_distribution(df, categorical_cols):
    """
    Analyzes the distribution of categorical variables.

    Args:
        df (pd.DataFrame): The input DataFrame.
        categorical_cols (list): List of categorical columns.

    Returns:
        dict: A dictionary containing value counts for each categorical column.
    """
    distributions = {}
    for col in categorical_cols:
        distributions[col] = df[col].value_counts()
    return distributions

def visualize_distribution(df, numerical_cols, categorical_cols):
    """
    Visualizes the distribution of numerical and categorical variables.

    Args:
        df (pd.DataFrame): The input DataFrame.
        numerical_cols (list): List of numerical columns.
        categorical_cols (list): List of categorical columns.
    """
    for col in numerical_cols:
        plt.figure(figsize=(8, 6))
        sns.histplot(df[col], kde=True)  # Use seaborn for better plots
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.show()

    for col in categorical_cols:
        plt.figure(figsize=(8, 6))
        sns.countplot(x=col, data=df)  # Use seaborn countplot
        plt.title(f'Distribution of {col}')
        plt.xlabel(col)
        plt.ylabel('Count')
        plt.show()

def analyze_sales_trends(df, date_col='Date', sales_col='Sales_Volume'):
    """
    Analyzes sales trends over time, including monthly and weekly trends.

    Args:
        df (pd.DataFrame): The input DataFrame.
        date_col (str): The name of the date column.
        sales_col (str): The name of the sales volume column.

    Returns:
        pd.DataFrame: DataFrame with aggregated sales data.
    """
    # Ensure the date column is correctly set.
    df[date_col] = pd.to_datetime(df[date_col])

    # Aggregate sales by month
    monthly_sales = df.groupby(pd.Grouper(key=date_col, freq='M'))[sales_col].sum().reset_index()
    monthly_sales['Month'] = monthly_sales[date_col].dt.strftime('%Y-%m')

    # Aggregate sales by week
    weekly_sales = df.groupby(pd.Grouper(key=date_col, freq='W'))[sales_col].sum().reset_index()
    weekly_sales['Week'] = weekly_sales[date_col].dt.strftime('%Y-W%U')

    # Aggregate sales by day of week
    daily_sales = df.groupby(df[date_col].dt.day_name())[sales_col].sum().reset_index()
    daily_sales = daily_sales.sort_values(by=sales_col, ascending=False)  # Sort by sales

    # Print the first few rows of the aggregated dataframes
    print("Monthly Sales Data:")
    print(monthly_sales.head())
    print("\nWeekly Sales Data:")
    print(weekly_sales.head())
    print("\nDaily Sales Data:")
    print(daily_sales.head())

    # Plotting
    plt.figure(figsize=(12, 6))
    sns.lineplot(x='Month', y=sales_col, data=monthly_sales)
    plt.title('Monthly Sales Trend')
    plt.xlabel('Month')
    plt.ylabel('Total Sales Volume')
    plt.xticks(rotation=45)
    plt.show()

    plt.figure(figsize=(12, 6))
    sns.lineplot(x='Week', y=sales_col, data=weekly_sales)
    plt.title('Weekly Sales Trend')
    plt.xlabel('Week')
    plt.ylabel('Total Sales Volume')
    plt.xticks(rotation=45)
    plt.show()

    plt.figure(figsize=(10, 6))
    sns.barplot(x=date_col, y=sales_col, data=daily_sales)
    plt.title('Daily Sales Trend')
    plt.xlabel('Day of Week')
    plt.ylabel('Total Sales Volume')
    plt.show()

    return monthly_sales, weekly_sales, daily_sales # Return the dataframes

def analyze_store_trends(df, store_col='Store_Location', sales_col='Sales_Volume', date_col='Date'):
    """
    Analyzes sales trends for each store.

    Args:
        df (pd.DataFrame): The input DataFrame.
        store_col (str): The name of the store location column.
        sales_col (str): The name of the sales volume column.
        date_col (str): The name of the date column.
    """
    # Ensure the date column is correctly set.
    df[date_col] = pd.to_datetime(df[date_col])
    store_sales = df.groupby([store_col, pd.Grouper(key=date_col, freq='M')])[sales_col].sum().reset_index()
    store_sales['Month'] = store_sales[date_col].dt.strftime('%Y-%m')

    # Print first few rows
    print("Store-wise Monthly Sales Data:")
    print(store_sales.head())

    plt.figure(figsize=(15, 7))
    sns.lineplot(x='Month', y=sales_col, hue=store_col, data=store_sales)
    plt.title('Monthly Sales Trend by Store')
    plt.xlabel('Month')
    plt.ylabel('Total Sales Volume')
    plt.xticks(rotation=45)
    plt.legend(title='Store Location')
    plt.show()

def analyze_category_trends(df, category_col='Product_Category', sales_col='Sales_Volume', date_col='Date'):
    """
    Analyzes sales trends for each product category.

    Args:
        df (pd.DataFrame): The input DataFrame.
        category_col (str): The name of the product category column.
        sales_col (str): The name of the sales volume column.
        date_col (str): The name of the date column.
    """
    # Ensure the date column is correctly set.
    df[date_col] = pd.to_datetime(df[date_col])
    category_sales = df.groupby([category_col, pd.Grouper(key=date_col, freq='M')])[sales_col].sum().reset_index()
    category_sales['Month'] = category_sales[date_col].dt.strftime('%Y-%m')

    # Print first few rows
    print("\nCategory-wise Monthly Sales Data:")
    print(category_sales.head())

    plt.figure(figsize=(15, 7))
    sns.lineplot(x='Month', y=sales_col, hue=category_col, data=category_sales)
    plt.title('Monthly Sales Trend by Category')
    plt.xlabel('Month')
    plt.ylabel('Total Sales Volume')
    plt.xticks(rotation=45)
    plt.legend(title='Product Category')
    plt.show()

def analyze_correlations(df, numerical_cols):
    """
    Analyzes correlations between numerical variables using a heatmap.

    Args:
        df (pd.DataFrame): The input DataFrame.
        numerical_cols (list): List of numerical columns.
    """
    correlation_matrix = df[numerical_cols].corr()
    plt.figure(figsize=(10, 8))
    sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
    plt.title('Correlation Heatmap')
    plt.show()

def detect_anomalies(df, sales_col='Sales_Volume', date_col='Date', threshold=3):
    """
    Detects anomalies in sales data using z-score.

    Args:
        df (pd.DataFrame): The input DataFrame.
        sales_col (str): The name of the sales volume column.
        date_col (str): The name of the date column.
        threshold (float): Z-score threshold for anomaly detection.
    Returns:
      pd.DataFrame: DataFrame containing the anomalies
    """
    df_copy = df.copy()
    z_scores = np.abs((df_copy[sales_col] - df_copy[sales_col].mean()) / df_copy[sales_col].std())
    anomalies = df_copy[z_scores > threshold]

    print("\nAnomalies Detected:")
    print(anomalies)

    if not anomalies.empty:
        plt.figure(figsize=(12, 6))
        sns.lineplot(x=date_col, y=sales_col, data=df_copy)
        plt.scatter(anomalies[date_col], anomalies[sales_col], color='red', label='Anomaly')
        plt.title('Sales Volume with Anomalies')
        plt.xlabel('Date')
        plt.ylabel('Sales Volume')
        plt.legend()
        plt.show()
    return anomalies

def generate_eda_report(df, numerical_cols, categorical_cols):
    """
    Generates a comprehensive EDA report.

    Args:
        df (pd.DataFrame): The input DataFrame.
        numerical_cols (list): List of numerical columns.
        categorical_cols (list): List of categorical columns.
    """
    print("\n\nExploratory Data Analysis Report\n" + "="*80)

    print("\n1. Descriptive Statistics\n" + "-"*80)
    print(get_descriptive_statistics(df, numerical_cols).to_string())

    print("\n\n2. Categorical Variable Distribution\n" + "-"*80)
    distributions = analyze_categorical_distribution(df, categorical_cols)
    for col, distribution in distributions.items():
        print(f"\nColumn: {col}\n{distribution.to_string()}\n")

    print("\n\n3. Data Distribution Visualization\n" + "-"*80)
    visualize_distribution(df, numerical_cols, categorical_cols)

    print("\n\n4. Sales Trend Analysis\n" + "-"*80)
    monthly_sales, weekly_sales, daily_sales = analyze_sales_trends(df) # Capture returned DataFrames.

    print("\n\n5. Store Trend Analysis\n" + "-"*80)
    analyze_store_trends(df)

    print("\n\n6. Category Trend Analysis\n" + "-"*80)
    analyze_category_trends(df)

    print("\n\n7. Correlation Analysis\n" + "-"*80)
    analyze_correlations(df, numerical_cols)

    '''print("\n\n8. Anomaly Detection\n" + "-"*80)
    detect_anomalies(df)'''

    print("\n\nEnd of EDA Report\n" + "="*80)

# Example usage:
numerical_cols = ['Sales_Volume', 'Price', 'Supplier_Cost', 'Replenishment_Lead_Time', 'Stock_Level']
categorical_cols = ['Product_Category', 'Store_Location', 'Promotion', 'Weekday'] # Removed Product_Name which has too many unique values
generate_eda_report(df, numerical_cols, categorical_cols)
```

```python
# --------------------------
# Long Short Term Memory (LSTM) Model
# --------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_squared_error, mean_absolute_error

# --------------------------
# Load the dataset from the specified path
# --------------------------
data_path = '/kaggle/input/super-store-sales-data-dataset/Super-Store-Sales-Data_.csv'
df_loaded = pd.read_csv(data_path, parse_dates=['Date'])
df_loaded = df_loaded.sort_values('Date')

# --------------------------
# Aggregate sales by date
# --------------------------
series = df_loaded.groupby('Date')['Sales_Volume'].sum().values.reshape(-1, 1)

# Check for NaN values
if np.isnan(series).any():
    print("NaN values detected, filling with zero")
    series = np.nan_to_num(series)

# --------------------------
# Normalize the data
# --------------------------
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_series = scaler.fit_transform(series)

# --------------------------
# Create dataset sequences with a fixed look_back period
# --------------------------
def create_dataset(dataset, look_back=30):
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        dataX.append(dataset[i:(i + look_back), 0])
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

look_back = 30
train_size = int(len(scaled_series) * 0.8)
train, test = scaled_series[:train_size], scaled_series[train_size:]
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, time steps, features]
trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 1))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 1))

# --------------------------
# Build and Train the LSTM Model
# --------------------------
model = Sequential()
model.add(LSTM(100, input_shape=(trainX.shape[1], 1)))
model.add(Dense(1))
model.compile(loss='mean_squared_error', optimizer='adam')

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(trainX, trainY, epochs=100, batch_size=64, verbose=1,
                    validation_split=0.1, callbacks=[early_stopping])

# --------------------------
# Evaluate the Model
# --------------------------
lstm_train_predict = model.predict(trainX)
lstm_test_predict = model.predict(testX)

# Inverse transform predictions and actual values
lstm_train_predict = scaler.inverse_transform(lstm_train_predict)
lstm_trainY_actual = scaler.inverse_transform(trainY.reshape(-1, 1))
lstm_test_predict = scaler.inverse_transform(lstm_test_predict)
lstm_testY_actual = scaler.inverse_transform(testY.reshape(-1, 1))

# Calculate RMSE & MAE
lstm_trainScoreRMSE = np.sqrt(mean_squared_error(lstm_trainY_actual, lstm_train_predict))
lstm_testScoreRMSE = np.sqrt(mean_squared_error(lstm_testY_actual, lstm_test_predict))
lstm_trainScoreMAE = mean_absolute_error(lstm_trainY_actual, lstm_train_predict)
lstm_testScoreMAE = mean_absolute_error(lstm_testY_actual, lstm_test_predict)

print(f'LSTM Train RMSE: {lstm_trainScoreRMSE:.2f}')
print(f'LSTM Test RMSE: {lstm_testScoreRMSE:.2f}')
print(f'LSTM Train MAE: {lstm_trainScoreMAE:.2f}')
print(f'LSTM Test MAE: {lstm_testScoreMAE:.2f}')

# --------------------------
# Visualize the Predictions
# --------------------------
lstm_trainPredictPlot = np.empty_like(series, dtype=float)
lstm_trainPredictPlot[:, :] = np.nan
lstm_trainPredictPlot[look_back:len(lstm_train_predict) + look_back, :] = lstm_train_predict

lstm_testPredictPlot = np.empty_like(series, dtype=float)
lstm_testPredictPlot[:, :] = np.nan
lstm_testPredictPlot[len(lstm_train_predict) + (look_back * 2):len(series), :] = lstm_test_predict

plt.figure(figsize=(14, 6))
plt.plot(series, label='Actual Sales')
plt.plot(lstm_trainPredictPlot, label='LSTM Train Predictions')
plt.plot(lstm_testPredictPlot, label='LSTM Test Predictions')
plt.xlabel('Days')
plt.ylabel('Sales Volume')
plt.title('Sales Prediction using LSTM')
plt.legend()
plt.show()
```

```python
import numpy as np

def predict_product_demand(df, model, scaler, look_back=30, days_ahead=7):
    """
    Predicts the total demand for 'Butter' for the next week using an LSTM model.
    
    :param df: DataFrame containing sales data with 'Date' and 'Product' columns
    :param model: Trained LSTM model
    :param scaler: MinMaxScaler used for normalization
    :param look_back: Number of previous days used for prediction
    :param days_ahead: Number of future days to predict (default: 7 for a week)
    :return: Total predicted sales volume for the next 7 days or 'Not Available' if the product is not found
    """
    
    product_name = "Butter"  # Fixed product name
    
    if product_name not in df['Product_Name'].unique():
        print("Not Available")
        return "Not Available"
    
    # Filter dataset for the specific product
    product_df = df[df['Product_Name'] == product_name].groupby('Date')['Sales_Volume'].sum()
    
    if product_df.empty:
        print("No historical data available for the product.")
        return "Not Available"
    
    product_series = product_df.values.reshape(-1, 1)
    
    # Scale the data
    scaled_series = scaler.fit_transform(product_series)
    
    # Ensure enough historical data is available
    if len(scaled_series) < look_back:
        print("Insufficient historical data for prediction.")
        return "Not Available"
    
    # Get the most recent sequence for prediction
    input_seq = scaled_series[-look_back:].reshape(1, look_back, 1)
    
    predictions = []
    for _ in range(days_ahead):
        next_pred = model.predict(input_seq, verbose=0)[0][0]
        predictions.append(next_pred)
        
        # Update input sequence with the new prediction
        input_seq = np.roll(input_seq, shift=-1, axis=1)
        input_seq[0, -1, 0] = next_pred
    
    # Inverse transform predictions to get actual sales values
    predictions = scaler.inverse_transform(np.array(predictions).reshape(-1, 1)).flatten()
    
    # Sum up the predictions for the total demand over the next week
    total_demand = np.sum(predictions)
    
    print(f"Total predicted demand for {product_name} in the next week: {total_demand:.2f}")
    return total_demand

# Example call to the function
predicted_demand = predict_product_demand(df, model, scaler)
```

```python
# --------------------------
# Convolutional Neural Networks (CNN) for Time Series Model
# --------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, Flatten, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.metrics import mean_squared_error, mean_absolute_error

# --------------------------
# Load the dataset from the specified path
# --------------------------
data_path = '/kaggle/input/super-store-sales-data-dataset/Super-Store-Sales-Data_.csv'
df_loaded = pd.read_csv(data_path, parse_dates=['Date'])
df_loaded = df_loaded.sort_values('Date')

# --------------------------
# Aggregate sales by date
# --------------------------
series = df_loaded.groupby('Date')['Sales_Volume'].sum().values.reshape(-1, 1)

# Check for NaN values and handle them
if np.isnan(series).any():
    print("NaN values detected, filling with zero")
    series = np.nan_to_num(series)

# --------------------------
# Normalize the data
# --------------------------
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_series = scaler.fit_transform(series)

# --------------------------
# Create dataset sequences with a fixed look_back period
# --------------------------
def create_dataset(dataset, look_back=30):
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        dataX.append(dataset[i:(i + look_back), 0])
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

look_back = 30
train_size = int(len(scaled_series) * 0.8)
train, test = scaled_series[:train_size], scaled_series[train_size:]
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, time steps, features]
trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 1))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 1))

# --------------------------
# Build and Train the CNN Model
# --------------------------
model = Sequential()

# First convolutional block
model.add(Conv1D(filters=64, kernel_size=3, activation='relu', input_shape=(trainX.shape[1], 1)))
model.add(Conv1D(filters=64, kernel_size=3, activation='relu'))
model.add(MaxPooling1D(pool_size=2))
model.add(Dropout(0.25))

# Flatten and add Dense layers
model.add(Flatten())
model.add(Dense(50, activation='relu'))
model.add(Dropout(0.5))
model.add(Dense(1))  # Output layer

model.compile(loss='mean_squared_error', optimizer='adam')

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True)
history = model.fit(trainX, trainY, epochs=100, batch_size=64, verbose=1,
                    validation_split=0.1, callbacks=[early_stopping])

# --------------------------
# Evaluate the Model
# --------------------------
cnn_trainPredict = model.predict(trainX)
cnn_testPredict = model.predict(testX)

# Inverse transform predictions and actual values
cnn_trainPredict = scaler.inverse_transform(cnn_trainPredict)
trainY_actual = scaler.inverse_transform(trainY.reshape(-1, 1))
cnn_testPredict = scaler.inverse_transform(cnn_testPredict)
testY_actual = scaler.inverse_transform(testY.reshape(-1, 1))

trainScoreRMSE = np.sqrt(mean_squared_error(trainY_actual, cnn_trainPredict))
testScoreRMSE = np.sqrt(mean_squared_error(testY_actual, cnn_testPredict))
trainScoreMAE = mean_absolute_error(trainY_actual, cnn_trainPredict)
testScoreMAE = mean_absolute_error(testY_actual, cnn_testPredict)

print(f'Train RMSE: {trainScoreRMSE:.2f}')
print(f'Test RMSE: {testScoreRMSE:.2f}')
print(f'Train MAE: {trainScoreMAE:.2f}')
print(f'Test MAE: {testScoreMAE:.2f}')

# --------------------------
# Visualize the Predictions
# --------------------------
trainPredictPlot = np.empty_like(series, dtype=float)
trainPredictPlot[:, :] = np.nan
trainPredictPlot[look_back:len(cnn_trainPredict) + look_back, :] = cnn_trainPredict

testPredictPlot = np.empty_like(series, dtype=float)
testPredictPlot[:, :] = np.nan
testPredictPlot[len(cnn_trainPredict) + (look_back * 2):len(series), :] = cnn_testPredict

plt.figure(figsize=(14, 6))
plt.plot(series, label='Actual Sales')
plt.plot(trainPredictPlot, label='Train Predictions')
plt.plot(testPredictPlot, label='Test Predictions')
plt.xlabel('Days')
plt.ylabel('Sales Volume')
plt.title('Sales Prediction using CNN')
plt.legend()
plt.show()
```

```python
# --------------------------
# Transformer-Based Models (Simplified Transformer Model)
# --------------------------
import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, LayerNormalization, MultiHeadAttention, Add
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

# --------------------------
# Load the dataset
# --------------------------
data_path = '/kaggle/input/super-store-sales-data-dataset/Super-Store-Sales-Data_.csv'
df_loaded = pd.read_csv(data_path, parse_dates=['Date'])
df_loaded = df_loaded.sort_values('Date')

# --------------------------
# Aggregate sales by date
# --------------------------
series = df_loaded.groupby('Date')['Sales_Volume'].sum().values.reshape(-1, 1)

# Handle NaN values
if np.isnan(series).any():
    print("NaN values detected, filling with zero")
    series = np.nan_to_num(series)

# --------------------------
# Normalize the data
# --------------------------
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_series = scaler.fit_transform(series)

# --------------------------
# Create dataset sequences
# --------------------------
def create_dataset(dataset, look_back=30):
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        dataX.append(dataset[i:(i + look_back), 0])
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

look_back = 30
train_size = int(len(scaled_series) * 0.8)
train, test = scaled_series[:train_size], scaled_series[train_size:]
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, time steps, features]
trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 1))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 1))

# --------------------------
# Transformer Model
# --------------------------

class TransformerBlock(tf.keras.layers.Layer):
    def __init__(self, embed_dim, num_heads, ff_dim, rate=0.1):
        super(TransformerBlock, self).__init__()
        self.att = MultiHeadAttention(num_heads=num_heads, key_dim=embed_dim)
        self.ffn = tf.keras.Sequential([
            Dense(ff_dim, activation="relu"),
            Dense(embed_dim),
        ])
        self.layernorm1 = LayerNormalization(epsilon=1e-6)
        self.layernorm2 = LayerNormalization(epsilon=1e-6)
        self.dropout1 = Dropout(rate)
        self.dropout2 = Dropout(rate)

    def call(self, inputs, training):
        attn_output = self.att(inputs, inputs)
        attn_output = self.dropout1(attn_output, training=training)
        out1 = self.layernorm1(inputs + attn_output)

        ffn_output = self.ffn(out1)
        ffn_output = self.dropout2(ffn_output, training=training)
        return self.layernorm2(out1 + ffn_output)

# Positional Encoding Function
def positional_encoding(seq_len, d_model):
    positions = np.arange(seq_len)[:, np.newaxis]
    angles = np.arange(d_model)[np.newaxis, :]
    angle_rates = 1 / np.power(10000, (2 * (angles // 2)) / np.float32(d_model))

    pos_encoding = positions * angle_rates
    pos_encoding[:, 0::2] = np.sin(pos_encoding[:, 0::2])
    pos_encoding[:, 1::2] = np.cos(pos_encoding[:, 1::2])
    
    return tf.cast(pos_encoding[np.newaxis, ...], dtype=tf.float32)

# Model Parameters
embed_dim = 64  # Embedding dimension
num_heads = 4   # Number of attention heads
ff_dim = 128    # Feed-forward layer dimension
dropout_rate = 0.1

# --------------------------
# Build Transformer Model
# --------------------------

inputs = Input(shape=(look_back, 1))
pos_encoding = positional_encoding(look_back, embed_dim)

x = Dense(embed_dim)(inputs) + pos_encoding  # Project input to embedding dimension

# 🔹 Pass `training=True` explicitly when calling TransformerBlock
x = TransformerBlock(embed_dim, num_heads, ff_dim, dropout_rate)(x, training=True)
x = TransformerBlock(embed_dim, num_heads, ff_dim, dropout_rate)(x, training=True)  # Stacking Transformers

x = tf.keras.layers.GlobalAveragePooling1D()(x)
x = Dense(64, activation="relu")(x)
x = Dropout(0.1)(x)
outputs = Dense(1)(x)

model = Model(inputs, outputs)
model.compile(optimizer=Adam(learning_rate=0.001), loss="mse")

# --------------------------
# Train the Model
# --------------------------
history = model.fit(trainX, trainY, epochs=100, batch_size=64, verbose=1, validation_split=0.1)

# --------------------------
# Make Predictions
# --------------------------
transformer_trainPredict = model.predict(trainX)
transformer_testPredict = model.predict(testX)

# Inverse transform predictions and actual values
transformer_trainPredict = scaler.inverse_transform(transformer_trainPredict)
transformer_trainY_actual = scaler.inverse_transform(trainY.reshape(-1, 1))
transformer_testPredict = scaler.inverse_transform(transformer_testPredict)
transformer_testY_actual = scaler.inverse_transform(testY.reshape(-1, 1))

transformer_trainScoreRMSE = np.sqrt(mean_squared_error(transformer_trainY_actual, transformer_trainPredict))
transformer_testScoreRMSE = np.sqrt(mean_squared_error(transformer_testY_actual, transformer_testPredict))
transformer_trainScoreMAE = mean_absolute_error(transformer_trainY_actual, transformer_trainPredict)
transformer_testScoreMAE = mean_absolute_error(transformer_testY_actual, transformer_testPredict)

print(f'Train RMSE: {transformer_trainScoreRMSE:.2f}')
print(f'Test RMSE: {transformer_testScoreRMSE:.2f}')
print(f'Train MAE: {transformer_trainScoreMAE:.2f}')
print(f'Test MAE: {transformer_testScoreMAE:.2f}')

# --------------------------
# Visualize the Predictions
# --------------------------
transformer_trainPredictPlot = np.empty_like(series, dtype=float)
transformer_trainPredictPlot[:, :] = np.nan
transformer_trainPredictPlot[look_back:len(transformer_trainPredict) + look_back, :] = transformer_trainPredict

transformer_testPredictPlot = np.empty_like(series, dtype=float)
transformer_testPredictPlot[:, :] = np.nan
transformer_testPredictPlot[len(transformer_trainPredict) + (look_back * 2):len(series), :] = transformer_testPredict

plt.figure(figsize=(14, 6))
plt.plot(series, label='Actual Sales')
plt.plot(trainPredictPlot, label='Train Predictions')
plt.plot(testPredictPlot, label='Test Predictions')
plt.xlabel('Days')
plt.ylabel('Sales Volume')
plt.title('Sales Prediction using Transformer Model')
plt.legend()
plt.show()
```

```python
# --------------------------
# Temporal Convolutional Networks (TCN) Model
# --------------------------
import pandas as pd
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout, Conv1D, BatchNormalization, ReLU, Add, Flatten
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error

# --------------------------
# Load the dataset
# --------------------------
data_path = '/kaggle/input/super-store-sales-data-dataset/Super-Store-Sales-Data_.csv'
df_loaded = pd.read_csv(data_path, parse_dates=['Date'])
df_loaded = df_loaded.sort_values('Date')

# --------------------------
# Aggregate sales by date
# --------------------------
series = df_loaded.groupby('Date')['Sales_Volume'].sum().values.reshape(-1, 1)

# Handle NaN values
if np.isnan(series).any():
    print("NaN values detected, filling with zero")
    series = np.nan_to_num(series)

# --------------------------
# Normalize the data
# --------------------------
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_series = scaler.fit_transform(series)

# --------------------------
# Create dataset sequences
# --------------------------
def create_dataset(dataset, look_back=30):
    dataX, dataY = [], []
    for i in range(len(dataset) - look_back):
        dataX.append(dataset[i:(i + look_back), 0])
        dataY.append(dataset[i + look_back, 0])
    return np.array(dataX), np.array(dataY)

look_back = 30
train_size = int(len(scaled_series) * 0.8)
train, test = scaled_series[:train_size], scaled_series[train_size:]
trainX, trainY = create_dataset(train, look_back)
testX, testY = create_dataset(test, look_back)

# Reshape input to [samples, time steps, features]
trainX = np.reshape(trainX, (trainX.shape[0], trainX.shape[1], 1))
testX = np.reshape(testX, (testX.shape[0], testX.shape[1], 1))

# --------------------------
# Temporal Convolutional Network (TCN)
# --------------------------

def residual_block(x, filters, dilation_rate, kernel_size=3, dropout_rate=0.2):
    """ A single residual block in TCN. """
    shortcut = x  # Residual connection

    # First Conv1D layer
    x = Conv1D(filters, kernel_size, dilation_rate=dilation_rate, padding="causal")(x)
    x = BatchNormalization()(x)
    x = ReLU()(x)
    x = Dropout(dropout_rate)(x)

    # Second Conv1D layer
    x = Conv1D(filters, kernel_size, dilation_rate=dilation_rate, padding="causal")(x)
    x = BatchNormalization()(x)
    x = ReLU()(x)
    x = Dropout(dropout_rate)(x)

    # Add residual connection
    if shortcut.shape[-1] != x.shape[-1]:  # Adjust dimensions if needed
        shortcut = Conv1D(filters, 1, padding="same")(shortcut)
    x = Add()([x, shortcut])

    return x

# --------------------------
# Build TCN Model
# --------------------------

inputs = Input(shape=(look_back, 1))

x = residual_block(inputs, filters=64, dilation_rate=1)
x = residual_block(x, filters=64, dilation_rate=2)
x = residual_block(x, filters=64, dilation_rate=4)

x = Flatten()(x)  # Flatten before dense layers
x = Dense(64, activation="relu")(x)
x = Dropout(0.1)(x)
outputs = Dense(1)(x)

model = Model(inputs, outputs)
model.compile(optimizer=Adam(learning_rate=0.001), loss="mse", metrics=['mae'])

# --------------------------
# Train the Model
# --------------------------
history = model.fit(trainX, trainY, epochs=100, batch_size=64, verbose=1, validation_split=0.1)

# --------------------------
# Make Predictions
# --------------------------
tcn_trainPredict = model.predict(trainX)
tcn_testPredict = model.predict(testX)

# Inverse transform predictions and actual values
tcn_trainPredict = scaler.inverse_transform(tcn_trainPredict)
tcn_trainY_actual = scaler.inverse_transform(trainY.reshape(-1, 1))
tcn_testPredict = scaler.inverse_transform(tcn_testPredict)
tcn_testY_actual = scaler.inverse_transform(testY.reshape(-1, 1))

# Compute Evaluation Metrics
tcn_trainScoreRMSE = np.sqrt(mean_squared_error(tcn_trainY_actual, tcn_trainPredict))
tcn_testScoreRMSE = np.sqrt(mean_squared_error(tcn_testY_actual, tcn_testPredict))
tcn_trainScoreMAE = mean_absolute_error(tcn_trainY_actual, tcn_trainPredict)
tcn_testScoreMAE = mean_absolute_error(tcn_testY_actual, tcn_testPredict)

print(f'TCN Train RMSE: {tcn_trainScoreRMSE:.2f}')
print(f'TCN Test RMSE: {tcn_testScoreRMSE:.2f}')
print(f'TCN Train MAE: {tcn_trainScoreMAE:.2f}')
print(f'TCN Test MAE: {tcn_testScoreMAE:.2f}')

# --------------------------
# Visualize the Predictions
# --------------------------
plt.figure(figsize=(14, 6))
plt.plot(series, label='Actual Sales')
plt.plot(np.arange(look_back, look_back + len(tcn_trainPredict)), tcn_trainPredict, label='TCN Train Predictions')
plt.plot(np.arange(len(series) - len(tcn_testPredict), len(series)), tcn_testPredict, label='TCN Test Predictions')
plt.xlabel('Days')
plt.ylabel('Sales Volume')
plt.title('Sales Prediction using Temporal Convolutional Network (TCN)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error

# --------------------------
# Function to Plot Predictions
# --------------------------
def plot_predictions(actual_train, actual_test, train_predictions, test_predictions, model_names, look_back):
    num_models = len(model_names)
    cols = 2  # Max 2 plots per row
    rows = (num_models + cols - 1) // cols  # Calculate required rows

    fig, axes = plt.subplots(rows, cols, figsize=(14, rows * 5))  # Bigger plots
    
    # Flatten axes if multiple rows
    axes = axes.flatten() if rows > 1 else [axes]

    for i, (train_pred, test_pred, name) in enumerate(zip(train_predictions, test_predictions, model_names)):
        predict_train_plot = np.full_like(actual_train, np.nan, dtype=float)  # Initialize array with NaN
        predict_test_plot = np.full_like(actual_test, np.nan, dtype=float)  # Initialize array with NaN
        
        predict_train_plot[:len(train_pred)] = train_pred.reshape(-1)  # Align train predictions
        predict_test_plot[:len(test_pred)] = test_pred.reshape(-1)  # Align test predictions
        
        ax = axes[i]
        ax.plot(actual_train, label='Actual Train Sales', linewidth=2, color='black')
        ax.plot(actual_test, label='Actual Test Sales', linewidth=2, linestyle='dotted', color='gray')
        ax.plot(predict_train_plot, label=f'{name} Train Predictions', linestyle='dashed')
        ax.plot(predict_test_plot, label=f'{name} Test Predictions', linestyle='dashdot')
        
        ax.set_xlabel('Days')
        ax.set_ylabel('Sales Volume')
        ax.set_title(f'Actual vs Predicted ({name})')
        ax.legend()
    
    # Remove extra empty subplots
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()

# --------------------------
# Function to Compute & Display Evaluation Metrics
# --------------------------
def display_metrics(actual_train, actual_test, train_predictions, test_predictions, model_names):
    metrics_data = []
    
    for train_pred, test_pred, name in zip(train_predictions, test_predictions, model_names):
        min_train_len = min(len(actual_train), len(train_pred))
        min_test_len = min(len(actual_test), len(test_pred))
        
        actual_train_trimmed = actual_train[:min_train_len]
        actual_test_trimmed = actual_test[:min_test_len]
        train_pred_trimmed = train_pred[:min_train_len]
        test_pred_trimmed = test_pred[:min_test_len]
        
        train_rmse = np.sqrt(mean_squared_error(actual_train_trimmed, train_pred_trimmed))
        train_mae = mean_absolute_error(actual_train_trimmed, train_pred_trimmed)
        test_rmse = np.sqrt(mean_squared_error(actual_test_trimmed, test_pred_trimmed))
        test_mae = mean_absolute_error(actual_test_trimmed, test_pred_trimmed)
        
        metrics_data.append([name, train_rmse, train_mae, test_rmse, test_mae])
    
    metrics_df = pd.DataFrame(metrics_data, columns=['Model', 'Train RMSE', 'Train MAE', 'Test RMSE', 'Test MAE'])
    print(metrics_df)

# --------------------------
# Prepare Data & Models List
# --------------------------
actual_train_series = series[:train_size].reshape(-1)  # Ensure 1D format
actual_test_series = series[train_size:].reshape(-1)  # Ensure 1D format

# Model Names (As Defined in Your Reference)
models = ['LSTM', 'Transformer', 'TCN', 'GRU', 'CNN', 'Hybrid CNN-LSTM']

# Ensure predictions are 1D & mapped to correct models
train_predictions = [
    lstm_train_predict.reshape(-1), 
    transformer_trainPredict.reshape(-1), 
    tcn_trainPredict.reshape(-1), 
    gru_train_predict.reshape(-1), 
    cnn_trainPredict.reshape(-1), 
    cnn_lstm_train_predict.reshape(-1)
]

test_predictions = [
    lstm_test_predict.reshape(-1), 
    transformer_testPredict.reshape(-1), 
    tcn_testPredict.reshape(-1), 
    gru_test_predict.reshape(-1), 
    cnn_testPredict.reshape(-1), 
    cnn_lstm_test_predict.reshape(-1)
]

# --------------------------
# Execute Visualizations & Metrics
# --------------------------
plot_predictions(actual_train_series, actual_test_series, train_predictions, test_predictions, models, look_back)
display_metrics(actual_train_series, actual_test_series, train_predictions, test_predictions, models)
```
